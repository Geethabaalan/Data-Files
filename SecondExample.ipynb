```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

print("Model loaded successfully!")
```

```python
def generate_output(prompt):
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        padding=True
    )

    outputs = model.generate(
        **inputs,
        max_length=100
    )

    result = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return result.strip()
```

```python
zero_shot_prompt = """
Classify the sentiment as Positive, Negative, or Neutral.

Sentence: The movie was absolutely fantastic!

Answer:
"""

output_zero = generate_output(zero_shot_prompt)

print("Zero-shot Prompt:")
print(zero_shot_prompt)

print("\nZero-shot Output:")
print(output_zero)
```

```python
one_shot_prompt = """
Classify sentiment.

Example:
Sentence: I love this product!
Sentiment: Positive

Now classify:
Sentence: The movie was absolutely fantastic!
Sentiment:
"""

output_one = generate_output(one_shot_prompt)

print("One-shot Prompt:")
print(one_shot_prompt)

print("\nOne-shot Output:")
print(output_one)
```

```python
few_shot_prompt = """
Classify sentiment.

Sentence: I love this phone!
Sentiment: Positive

Sentence: This is the worst experience ever.
Sentiment: Negative

Sentence: The service was okay.
Sentiment: Neutral

Now classify:
Sentence: The movie was absolutely fantastic!
Sentiment:
"""

output_few = generate_output(few_shot_prompt)

print("Few-shot Prompt:")
print(few_shot_prompt)

print("\nFew-shot Output:")
print(output_few)
```

```python
print("----- FINAL COMPARISON -----")

print("Zero-shot :", output_zero)
print("One-shot  :", output_one)
print("Few-shot  :", output_few)
```
