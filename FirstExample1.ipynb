```python
#Chatbot with conversation memory
```

```python
from openai import OpenAI

client = OpenAI(api_key=api_key)

conversation = []

print("AI Chatbot")
print("Type 'exit' to stop.\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        break

    conversation.append({
        "role": "user",
        "content": user_input
    })

    response = client.responses.create(
        model="gpt-5.6",
        input=conversation
    )

    answer = response.output_text

    print("Bot:", answer)

    conversation.append({
        "role": "assistant",
        "content": answer
    })
```

```python
#Simple chatbot with system message
```

```python
from openai import OpenAI

client = OpenAI(api_key=api_key)

system_message = """
You are a helpful AI teaching assistant.
Explain concepts in simple language.
Give examples whenever possible.
If the user asks a technical question, provide Python examples.
"""

print("AI Chatbot")
print("Type 'exit' to stop.\n")

conversation = []

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    conversation.append({
        "role": "user",
        "content": user_input
    })

    response = client.responses.create(
        model="gpt-5.6",
        instructions=system_message,
        input=conversation
    )

    answer = response.output_text

    print("Chatbot:", answer)

    conversation.append({
        "role": "assistant",
        "content": answer
    })
```
